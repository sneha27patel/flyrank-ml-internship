## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Load dataset safely
try:
    df = pd.read_csv('data/raw/content_refresh_anonymized.csv')
except Exception:
    df = pd.read_csv('https://raw.githubusercontent.com/flyrank-bih/flyrank-ml-internship-starter/main/data/raw/content_refresh_anonymized.csv')

# 1. Total content pages
print("1. Total content pages in dataset:", len(df))

# 2. Pages currently in a 'down' trend
decline_count = (df['trend_direction'] == 'down').sum()
print(f"2. Pages in decline ('down' trend): {decline_count} ({decline_count/len(df)*100:.1f}%)")

# 3. Average position comparison
avg_pos_down = df[df['trend_direction'] == 'down']['avg_position'].mean()
avg_pos_up = df[df['trend_direction'] == 'up']['avg_position'].mean()
print(f"3. Average Position — Declining pages: {avg_pos_down:.1f} | Growing pages: {avg_pos_up:.1f}")
```
